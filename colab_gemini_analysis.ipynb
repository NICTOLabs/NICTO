# NICTO AI - Send Code to Gemini
This notebook loads the NICTO codebase and sends it to Gemini for analysis.

**Setup:**
1. Runtime -> Change runtime type -> T4 GPU
2. Upload the NICTO folder to /content/ using the file browser
3. Run all cells
4. Gemini will analyze the code and suggest training improvements

````
# Step 2: Set up API key
# Option A: Use Colab Secrets (recommended)
# Click the 🔑 icon on the left sidebar
# Add: Name=GOOGLE_API_KEY, Value=your_key
# Get key from: https://aistudio.google.com/app/apikey

from google.colab import userdata
import os

try:
    API_KEY = userdata.get('GOOGLE_API_KEY')
    print("API key loaded from Colab Secrets")
except:
    API_KEY = input("Enter your Gemini API key: ")
    print("Using manually entered key")

os.environ['GOOGLE_API_KEY'] = API_KEY
````

````
# Step 3: Load NICTO codebase
import glob

NICTO_DIR = "/content/NICTO/nicto_ai"  # Adjust if needed

code_files = {}
for filepath in sorted(glob.glob(f"{NICTO_DIR}/**/*.py", recursive=True)):
    if "__pycache__" in filepath:
        continue
    try:
        with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
            content = f.read()
        if len(content) < 50000:  # Skip huge files
            code_files[filepath] = content
            print(f"Loaded: {filepath.replace(NICTO_DIR + '/', '')}")
    except:
        pass

print(f"\nLoaded {len(code_files)} files")
print(f"Total: {sum(len(v) for v in code_files.values()):,} chars")
````

````
# Step 4: Send to Gemini for analysis
import google.generativeai as genai

genai.configure(api_key=API_KEY)
model = genai.GenerativeModel('gemini-2.5-flash')

# Build context
code_context = "\n\n".join([
    f"=== FILE: {fp.replace(NICTO_DIR + '/', '')} ===\n```python\n{content}\n```"
    for fp, content in code_files.items()
])

prompt = f"""You are analyzing the NICTO AI codebase ({len(code_files)} files).

{code_context}

TASKS:
1. Verify all components work (no mocks/stubs)
2. Find bugs
3. Suggest max model size for T4 GPU (16GB VRAM)
4. Write a complete Colab training script
5. Recommend training data

Give me working code I can run immediately."""

print("Sending to Gemini...")
response = model.generate_content(prompt)
print("\n" + response.text)
````

````
# Step 5: Save response
with open("gemini_analysis.md", "w") as f:
    f.write("# NICTO Analysis by Gemini\n\n" + response.text)
print("Saved to gemini_analysis.md")

# Download
from google.colab import files
files.download("gemini_analysis.md")
````
